```python
# Standardize OS values
df["device_os"] = (
    df["device_os"]
    .astype(str)
    .str.strip()
    .replace({
        "ios": "iOS",
        "IOS": "iOS",
        "android": "Android"
    })
)
```

```python
# check activities
df["activity"].value_counts()
```

```python
# Clean text columns
text_columns = [
    "device_id",
    "persona",
    "activity",
    "device_os",
    "ping_id",
    "day_of_week"
]

for column in text_columns:
    df[column] = (
        df[column]
        .astype(str)
        .str.strip()
    )
```

```python
# Convert GPS coordinates to numeric
df["lat"] = pd.to_numeric(
    df["lat"],
    errors="coerce"
)

df["lon"] = pd.to_numeric(
    df["lon"],
    errors="coerce"
)
```

```python
df[["lat", "lon"]].dtypes
```

```python
# Convert GPS accuracy
df["gps_accuracy_m"] = pd.to_numeric(
    df["gps_accuracy_m"],
    errors="coerce"
)
```

```python
# Convert distance from home
df["distance_from_home_m"] = pd.to_numeric(
    df["distance_from_home_m"],
    errors="coerce"
)
```

```python
# Convert sequence number
df["sequence_no"] = pd.to_numeric(
    df["sequence_no"],
    errors="coerce"
)
```

```python
# Convert timestamp
df["timestamp"] = pd.to_datetime(
    df["timestamp"],
    errors="coerce"
)
```

```python
df["timestamp"].dtype
```

```python
# Convert date
df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce"
).dt.date
```

```python
# Check invalid timestamps
invalid_timestamp_count = df["timestamp"].isna().sum()

print(
    "Invalid timestamps:",
    invalid_timestamp_count
)
```

```python
# Check latitude validity
invalid_lat = (
    (df["lat"] < -90) |
    (df["lat"] > 90) |
    df["lat"].isna()
)

print(
    "Invalid latitude records:",
    invalid_lat.sum()
)
```

```python
# Check longitude validity
invalid_lon = (
    (df["lon"] < -180) |
    (df["lon"] > 180) |
    df["lon"].isna()
)

print(
    "Invalid longitude records:",
    invalid_lon.sum()
)
```

```python
# Remove invalid GPS records
df = df[
    ~invalid_lat &
    ~invalid_lon
].copy()

df.reset_index(drop=True, inplace=True)

print("Dataset shape after GPS validation:", df.shape)
```

```python
# Validate GPS accuracy

print(
    "Minimum GPS accuracy:",
    df["gps_accuracy_m"].min()
)

print(
    "Maximum GPS accuracy:",
    df["gps_accuracy_m"].max()
)
```

```python
invalid_accuracy = (
    (df["gps_accuracy_m"] <= 0) |
    df["gps_accuracy_m"].isna()
)

print(
    "Invalid GPS accuracy records:",
    invalid_accuracy.sum()
)
```

```python
# Validate distance from home

invalid_distance = (
    (df["distance_from_home_m"] < 0) |
    df["distance_from_home_m"].isna()
)

print(
    "Invalid distance records:",
    invalid_distance.sum()
)
```

```python
# Validate sequence numbers

print(
    "Minimum sequence number:",
    df["sequence_no"].min()
)

print(
    "Maximum sequence number:",
    df["sequence_no"].max()
)
```

```python
# checking missing sequence values
print(
    "Missing sequence numbers:",
    df["sequence_no"].isna().sum()
)
```

```python
# Validate ping IDs

print(
    "Total rows:",
    len(df)
)

print(
    "Unique ping IDs:",
    df["ping_id"].nunique()
)
```

```python
duplicate_ping_ids = (
    df["ping_id"].duplicated().sum()
)

print(
    "Duplicate ping IDs:",
    duplicate_ping_ids
)
```

```python
# Check duplicate device/timestamp combinations

duplicate_device_time = df.duplicated(
    subset=[
        "device_id",
        "timestamp"
    ]
).sum()

print(
    "Duplicate device-timestamp records:",
    duplicate_device_time
)
```

```python
# Sort the dataset

df = df.sort_values(
    by=[
        "device_id",
        "timestamp",
        "sequence_no"
    ]
).reset_index(drop=True)
```

```python
# Recalculate temporal features

df["date"] = df["timestamp"].dt.date

df["hour"] = df["timestamp"].dt.hour

df["minute"] = df["timestamp"].dt.minute

df["day_of_week"] = (
    df["timestamp"]
    .dt.day_name()
)
```

```python
# Create time period

def get_time_period(hour):

    if 7 <= hour < 10:
        return "Morning"

    elif 10 <= hour < 16:
        return "Afternoon"

    elif 16 <= hour < 20:
        return "Evening"

    else:
        return "Night"


df["time_period"] = (
    df["hour"]
    .apply(get_time_period)
)
```

```python
df["time_period"].value_counts()
```

```python
# Final missing-value check

final_missing = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().sum() /
        len(df) *
        100
    ).round(2)
})

final_missing
```

```python
# Final duplicate check

print(
    "Final duplicate rows:",
    df.duplicated().sum()
)

print(
    "Final duplicate ping IDs:",
    df["ping_id"].duplicated().sum()
)
```
