```python
import pandas as pd
import numpy as np
import os
```

```python
# define path
VISITS_INPUT = r"C:\Users\Dell\Desktop\Geopulse\store_visits.csv"

SUMMARY_INPUT = r"C:\Users\Dell\Desktop\Geopulse\footfall_summary.csv"

print("Store visits input:", VISITS_INPUT)
print("Footfall summary input:", SUMMARY_INPUT)
```

```python
# Check files

print(
    "store_visits.csv exists:",
    os.path.exists(VISITS_INPUT)
)

print(
    "footfall_summary.csv exists:",
    os.path.exists(SUMMARY_INPUT)
)
```

```python
# Load store_visits.csv

visits = pd.read_csv(VISITS_INPUT)

print("Store visits loaded successfully!")

print("Rows:", visits.shape[0])
print("Columns:", visits.shape[1])
```

```python
# Understand the store-visits structure

visits.head(10)
```

```python
# Column names

for i, column in enumerate(
    visits.columns,
    start=1
):
    print(i, "→", column)
```

```python
# data types

visits.dtypes
```

```python
# dataset information

visits.info()
```

```python
# statistical summary

visits.describe()
```

```python
# Check missing values

missing = pd.DataFrame({
    "Missing_Count": visits.isnull().sum(),
    "Missing_Percentage": (
        visits.isnull().sum()
        / len(visits)
        * 100
    ).round(2)
})

missing
```

```python
# Check duplicate records

print(
    "Duplicate complete rows:",
    visits.duplicated().sum()
)
```

```python
# check duplicate visit IDs
print(
    "Total visits:",
    len(visits)
)

print(
    "Unique visit IDs:",
    visits["visit_id"].nunique()
)

print(
    "Duplicate visit IDs:",
    visits["visit_id"].duplicated().sum()
)
```

```python
# Clean text columns

text_columns = [
    "visit_id",
    "device_id",
    "persona",
    "store_id",
    "store_name",
    "category",
    "day_of_week"
]

for column in text_columns:

    visits[column] = (
        visits[column]
        .astype("string")
        .str.strip()
    )
```

```python
# Convert timestamps

visits["entry_time"] = pd.to_datetime(
    visits["entry_time"],
    errors="coerce"
)

visits["exit_time"] = pd.to_datetime(
    visits["exit_time"],
    errors="coerce"
)
```

```python
# Check timestamp problems

print(
    "Invalid entry times:",
    visits["entry_time"].isna().sum()
)

print(
    "Invalid exit times:",
    visits["exit_time"].isna().sum()
)
```

```python
# Validate entry and exit time

invalid_times = (
    visits["exit_time"]
    < visits["entry_time"]
)

print(
    "Invalid visit time records:",
    invalid_times.sum()
)
```

```python
# Validate dwell time

print(
    visits["dwell_minutes"].describe()
)
```

```python
# check negative values

invalid_dwell = (
    visits["dwell_minutes"] < 0
)

print(
    "Negative dwell times:",
    invalid_dwell.sum()
)
```

```python
# checking minimum

print(
    "Minimum dwell:",
    visits["dwell_minutes"].min()
)

print(
    "Minimum required dwell:",
    5
)
```

```python
# Recalculate dwell time

calculated_dwell = (
    visits["exit_time"]
    - visits["entry_time"]
).dt.total_seconds() / 60

visits["calculated_dwell_minutes"] = (
    calculated_dwell.round(1)
)
```

```python
visits[
    [
        "dwell_minutes",
        "calculated_dwell_minutes"
    ]
].head(10)
```

```python
# Find dwell-time inconsistencies

dwell_difference = (
    visits["dwell_minutes"]
    - visits["calculated_dwell_minutes"]
).abs()

print(
    "Maximum dwell difference:",
    dwell_difference.max()
)
```

```python
# Remove temporary column

visits.drop(
    columns=["calculated_dwell_minutes"],
    inplace=True
)
```

```python
# Validate ping count

print(
    visits["ping_count"].describe()
)
```

```python
invalid_ping_count = (
    visits["ping_count"] <= 0
)

print(
    "Invalid ping counts:",
    invalid_ping_count.sum()
)
```

```python
# Validate store IDs

print(
    visits["store_id"].value_counts()
)
```

```python
valid_store_ids = {
    "STR001",
    "STR002",
    "STR003"
}

invalid_stores = ~visits["store_id"].isin(
    valid_store_ids
)

print(
    "Invalid store IDs:",
    invalid_stores.sum()
)
```

```python
# Validate personas

print(
    visits["persona"].value_counts()
)
```

```python
# Recreate date from entry time

visits["date"] = (
    visits["entry_time"]
    .dt.date
)
```

```python
# Recreate day of week

visits["day_of_week"] = (
    visits["entry_time"]
    .dt.day_name()
)
```

```python
# Final store-visits quality check

print("Rows:", len(visits))
print("Columns:", len(visits.columns))

print(
    "\nMissing values:"
)

print(
    visits.isnull().sum()
)

print(
    "\nDuplicate rows:",
    visits.duplicated().sum()
)

print(
    "\nDuplicate visit IDs:",
    visits["visit_id"].duplicated().sum()
)
```

```python
# Sort store visits

visits = visits.sort_values(
    by=[
        "store_id",
        "entry_time",
        "device_id"
    ]
).reset_index(drop=True)
```

```python
# Save cleaned store visits

os.makedirs(
    "data/processed",
    exist_ok=True
)

visits.to_csv(
    VISITS_OUTPUT,
    index=False
)

print(
    "Clean store visit dataset saved!"
)

print(
    VISITS_OUTPUT
)
```

```python
# footfall_summary.csv
# Load summary

summary = pd.read_csv(
    SUMMARY_INPUT
)

print(
    "Footfall summary loaded!"
)

print(
    "Rows:",
    summary.shape[0]
)

print(
    "Columns:",
    summary.shape[1]
)
```

```python
# Inspect structure

summary.head()

```

```python
summary.info()
```

```python
summary.dtypes
```

```python
# Check missing values

summary_missing = pd.DataFrame({
    "Missing_Count": summary.isnull().sum(),

    "Missing_Percentage": (
        summary.isnull().sum()
        / len(summary)
        * 100
    ).round(2)
})

summary_missing
```

```python
# Check duplicates

print(
    "Duplicate rows:",
    summary.duplicated().sum()
)
```

```python
duplicate_store_date = summary.duplicated(
    subset=[
        "store_id",
        "date"
    ]
).sum()

print(
    "Duplicate store-date combinations:",
    duplicate_store_date
)
```

```python
# Clean summary text fields

summary_text_columns = [
    "store_id",
    "store_name",
    "category"
]

for column in summary_text_columns:

    summary[column] = (
        summary[column]
        .astype("string")
        .str.strip()
    )
```

```python
# Convert summary date

summary["date"] = pd.to_datetime(
    summary["date"],
    errors="coerce"
)
```

```python
print(
    "Invalid dates:",
    summary["date"].isna().sum()
)
```

```python
# Convert numerical columns

numeric_columns = [
    "total_visits",
    "unique_visitors",
    "total_dwell_minutes",
    "avg_dwell_minutes"
]

for column in numeric_columns:

    summary[column] = pd.to_numeric(
        summary[column],
        errors="coerce"
    )
```

```python
# Check summary statistics

summary[
    numeric_columns
].describe()
```

```python
# Validate total visits

invalid_visits = (
    summary["total_visits"] <= 0
)

print(
    "Invalid total visits:",
    invalid_visits.sum()
)
```

```python
# Validate unique visitors

invalid_unique_visitors = (
    (summary["unique_visitors"] <= 0)
    |
    (
        summary["unique_visitors"]
        > summary["total_visits"]
    )
)

print(
    "Invalid unique visitor records:",
    invalid_unique_visitors.sum()
)
```

```python
# Validate dwell time

invalid_total_dwell = (
    summary["total_dwell_minutes"] < 0
)

invalid_avg_dwell = (
    summary["avg_dwell_minutes"] < 0
)

print(
    "Invalid total dwell:",
    invalid_total_dwell.sum()
)

print(
    "Invalid average dwell:",
    invalid_avg_dwell.sum()
)
```

```python
# Cross-check average dwell

calculated_avg = (
    summary["total_dwell_minutes"]
    / summary["total_visits"]
).round(1)

summary["calculated_avg_dwell"] = calculated_avg
```

```python
summary[
    [
        "avg_dwell_minutes",
        "calculated_avg_dwell"
    ]
]
```

```python
# Check differences

avg_difference = (
    summary["avg_dwell_minutes"]
    - summary["calculated_avg_dwell"]
).abs()

print(
    "Maximum average dwell difference:",
    avg_difference.max()
)
```

```python
summary.drop(
    columns=["calculated_avg_dwell"],
    inplace=True
)
```

```python
# Validate store IDs

invalid_summary_stores = (
    ~summary["store_id"].isin(
        [
            "STR001",
            "STR002",
            "STR003"
        ]
    )
)

print(
    "Invalid store IDs:",
    invalid_summary_stores.sum()
)
```

```python
# Final summary quality check

print(
    "Rows:",
    len(summary)
)

print(
    "\nMissing values:"
)

print(
    summary.isnull().sum()
)

print(
    "\nDuplicate rows:",
    summary.duplicated().sum()
)

print(
    "\nDuplicate store-date combinations:",
    summary.duplicated(
        subset=[
            "store_id",
            "date"
        ]
    ).sum()
)
```

```python
# Save cleaned summary

summary = summary.sort_values(
    by=[
        "date",
        "store_id"
    ]
).reset_index(drop=True)

summary.to_csv(
    SUMMARY_OUTPUT,
    index=False
)

print(
    "Clean footfall summary saved!"
)

print(
    SUMMARY_OUTPUT
)
```
