```python
# check GPS data types
data["lat"] = pd.to_numeric(
    data["lat"],
    errors="coerce"
)

data["lon"] = pd.to_numeric(
    data["lon"],
    errors="coerce"
)
```

```python
data[["lat", "lon"]].dtypes
```

```python
# checking latitude
invalid_lat = (
    (data["lat"] < -90) |
    (data["lat"] > 90)
)

print(
    "Invalid latitude records:",
    invalid_lat.sum()
)
```

```python
# checking longitude
invalid_lon = (
    (data["lon"] < -180) |
    (data["lon"] > 180)
)

print(
    "Invalid longitude records:",
    invalid_lon.sum()
)
```

```python
# Checking simulated study area
MIN_LAT = 20.95
MAX_LAT = 21.25

MIN_LON = 78.90
MAX_LON = 79.20
```

```python
outside_area = (
    (data["lat"] < MIN_LAT) |
    (data["lat"] > MAX_LAT) |
    (data["lon"] < MIN_LON) |
    (data["lon"] > MAX_LON)
)

print(
    "Records outside study area:",
    outside_area.sum()
)
```

```python
# checking devices
unique_devices = data["device_id"].nunique()

print(
    "Unique devices:",
    unique_devices
)
```

```python
# Checking personas
data["persona"].value_counts()
```

```python
# percentage
(
    data["persona"]
    .value_counts(normalize=True)
    * 100
).round(2)
```

```python
# checking activities
data["activity"].value_counts()
```

```python
# Clean text fields
data["persona"] = (
    data["persona"]
    .astype(str)
    .str.strip()
)

data["activity"] = (
    data["activity"]
    .astype(str)
    .str.strip()
)
```

```python
# Sort the GPS records
data = data.sort_values(
    by=["device_id", "timestamp"]
).reset_index(drop=True)
```

```python
# Now FEATURE ENGINEERING
# This isn't strictly "cleaning", but it's an important next stage of your data preparation.
# Create date

data["date"] = data["timestamp"].dt.date
```

```python
# Create hour
data["hour"] = data["timestamp"].dt.hour
```

```python
# Create day of week

data["day_of_week"] = (
    data["timestamp"]
    .dt.day_name()
)
```

```python
# Create time period

def get_time_period(hour):

    if 7 <= hour < 10:
        return "Morning"

    elif 10 <= hour < 16:
        return "Afternoon"

    elif 16 <= hour < 20:
        return "Evening"

    else:
        return "Night"


data["time_period"] = (
    data["hour"]
    .apply(get_time_period)
)
```
